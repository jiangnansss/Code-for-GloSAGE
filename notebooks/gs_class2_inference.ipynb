# GS_class2 model comparison
Provide one PAS-GS fine-tuned binary checkpoint per model. Pretrained foundation-model weights alone are not classifiers.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd().resolve()
if not (root / 'evaluate_gs.py').exists():
    root = root.parent
assert (root / 'evaluate_gs.py').exists(), 'Open notebook from the repository or notebooks directory'
data = root / 'data/GS_class2'
weights = root / 'weights'
checkpoints = {name: weights / f'{name}_pas_gs.pth' for name in ['glosage', 'glopath', 'uni', 'plip', 'conch', 'keep']}
keep_config = weights / 'KEEP/config.json'
for name, path in checkpoints.items():
    if not path.is_file(): raise FileNotFoundError(f'{name}: {path}')
cmd = [sys.executable, str(root / 'evaluate_gs.py'), '--data', str(data), '--keep-config', str(keep_config), '--output', str(root / 'outputs/gs_class2_metrics.csv')]
for name, path in checkpoints.items():
    cmd += ['--checkpoint', f'{name}={path}']
subprocess.run(cmd, cwd=root, check=True)
import pandas as pd
display(pd.read_csv(root / 'outputs/gs_class2_metrics.csv'))
```
